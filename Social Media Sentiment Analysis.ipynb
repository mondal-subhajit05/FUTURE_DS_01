DATA CLEANING & TRANSFORMATION

```python
# Check for missing values
df.isnull().sum()
```

```python
#Delete unnecessary columns
df=df.drop(columns=["Unnamed: 0.1", "Unnamed: 0"])
```

```python
df.columns.tolist()
```

```python
#Count of each sentiment type
df["Sentiment"].value_counts()
```

```python
#Filter positive sentiment tweets
positive_tweets=df[df["Sentiment"].str.strip() == "Positive"]
positive_tweets.head()
```

```python
#Filter negative sentiment tweets
negative_tweets=df[df["Sentiment"].str.strip() == "Negative"]
negative_tweets.head()
```

```python
#Filter neutral sentiment tweets
neutral_tweets=df[df["Sentiment"].str.strip() == "Neutral"]
neutral_tweets.head()
```

```python
#Analyze top hashtags used
hashtags=df["Hashtags"].dropna().str.split()
hashtags_flat=[tag for sublist in hashtags for tag in sublist]
hashtags_counter=Counter(hashtags_flat)
# Display top 10 hashtags
hashtags_counter.most_common(10)
```

```python
#Find the most liked posts
df.nlargest(10, "Likes")[["Text", "Likes"]]
```

```python
#Average Likes and Retweets per Sentiment
df.groupby("Sentiment")[["Likes", "Retweets"]].mean()
```

```python
#Top 5 Users with Most Posts
df["User"].value_counts().head(5)
```

```python
#Most Used Platforms
df["Platform"].value_counts()
```

```python
#Sentiment Trends Over Months
df.groupby(["Year", "Month"])["Sentiment"].value_counts().unstack()
```
